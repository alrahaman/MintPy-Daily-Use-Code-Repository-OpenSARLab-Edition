4 — Locate the Processing Reference Pixel Indices
What it does: Retrieves the row/column grid index that MintPy used as its zero-movement anchor during time-series inversion.
When to use: Always verify this after processing. If the reference pixel landed on water, a construction site, or a noisy corner, your entire velocity map is biased.

```python
from mintpy.utils import readfile

file_path = f"{DATA_DIR}/velocity.h5"
meta = readfile.read_attribute(file_path)

ref_y = int(meta['REF_Y'])
ref_x = int(meta['REF_X'])
ref_lat = meta.get('REF_LAT', 'N/A')
ref_lon = meta.get('REF_LON', 'N/A')

print(f"Reference pixel → Row: {ref_y}, Col: {ref_x}")
print(f"Geographic location  → Lat: {ref_lat}, Lon: {ref_lon}")
```
