3 — Lightweight Single Attribute Query
What it does: Reads only the metadata dictionary without loading any pixel data into RAM. Returns a single metadata value by key.
When to use: When you just need a quick fact — like the reference point coordinates, satellite heading, or wavelength — without waiting for a gigabyte file to load.

```python
import os
from mintpy.utils import readfile

# 1. Redefine the variable so Python knows what it is
DATA_DIR = "/home/jovyan/dhaka/HyP3_downloads/MintPy" 

# 2. Build the file path
file_path = f"{DATA_DIR}/velocity.h5"

# 3. Read the attributes safely
meta = readfile.read_attribute(file_path)

ref_lat = meta.get('REF_LAT', 'N/A')
ref_lon = meta.get('REF_LON', 'N/A')
wavelength = meta.get('WAVELENGTH', 'N/A')
platform = meta.get('PLATFORM', 'N/A')

print(f"Reference point : Lat {ref_lat}, Lon {ref_lon}")
print(f"Wavelength (m)  : {wavelength}")
print(f"Platform        : {platform}")
```
