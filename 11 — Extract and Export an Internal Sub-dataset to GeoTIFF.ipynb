11 — Extract and Export an Internal Sub-dataset to GeoTIFF
What it does: Reaches inside a multi-layer MintPy HDF5 file and pulls out one specific internal dataset (like the velocity standard deviation error map) to save as a separate GeoTIFF.
When to use: When you need to isolate and share uncertainty or error layers for independent quality assessment.
Note: These scripts read data from DATA_DIR but save output files into the folder where your .ipynb notebook currently lives, unless you specify an absolute output path.


```python
from mintpy.cli import save_gdal

file_path = f"{DATA_DIR}/velocity.h5"
output_path = f"{DATA_DIR}/velocity_error.tif"

# -d specifies the internal HDF5 dataset name to extract
save_gdal.main([file_path, '-d', 'velocityStd', '-o', output_path])
print(f"✅ Error layer saved to: {output_path}")
```
