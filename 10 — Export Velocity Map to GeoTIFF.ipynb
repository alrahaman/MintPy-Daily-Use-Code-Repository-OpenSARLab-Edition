10 — Export Velocity Map to GeoTIFF
What it does: Converts the HDF5 velocity file into a standard .tif GeoTIFF readable by QGIS, ArcGIS, GDAL, and GeoPandas.
When to use: When sharing data with colleagues who don't use MintPy/Python, or when you need to do raster operations in another tool.
Note: These scripts read data from DATA_DIR but save output files into the folder where your .ipynb notebook currently lives, unless you specify an absolute output path.

```python
from mintpy.cli import save_gdal

file_path = f"{DATA_DIR}/velocity.h5"
output_path = f"{DATA_DIR}/velocity_export.tif"

save_gdal.main([file_path, '-o', output_path, '--of', 'GTiff'])
print(f"✅ GeoTIFF saved to: {output_path}")
```
