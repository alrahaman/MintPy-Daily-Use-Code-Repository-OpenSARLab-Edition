2 — Read a 3D Time-Series Array and its Date List
What it does: Loads the full 3D displacement cube (time, rows, cols) and the chronological list of SAR acquisition dates into memory.
When to use: Starting point for any time-series statistical analysis, seasonal decomposition, or trend fitting.

```python
import os
from mintpy.objects import timeseries

file_path = f"{DATA_DIR}/timeseries_ERA5_ramp_demErr.h5"

# FIX: Initialize and open the file as a timeseries object
ts_obj = timeseries(file_path)
ts_obj.open()

# Now you can easily call its built-in attributes
date_list = ts_obj.dateList
ts_data = ts_obj.read()  # if you still need the data matrix

print(f"3D cube shape          : {ts_data.shape}  →  (dates, rows, cols)")
print(f"Total acquisition steps: {len(date_list)}")
print(f"First acquisition date : {date_list[0]}")
print(f"Last  acquisition date : {date_list[-1]}")
print(f"Unit in file           : {ts_obj.metadata.get('UNIT', 'unknown')}")
```
