1 — Read a 2D Dataset (Velocity or Coherence Map)
What it does: Loads a static 2D raster (e.g. velocity, coherence) into a NumPy array and returns the associated metadata dictionary for that file.
When to use: When you need the raw matrix and its spatial parameters for custom processing or statistics.

```python
from mintpy.utils import readfile

file_path = f"{DATA_DIR}/velocity.h5"
vel_data, vel_meta = readfile.read(file_path)

print("Matrix shape (rows, cols):", vel_data.shape)
print("Data type               :", vel_data.dtype)
print("EPSG / Coordinate system:", vel_meta.get('EPSG', 'Not specified'))
print("X pixel size (degrees)  :", vel_meta.get('X_STEP'))
print("Y pixel size (degrees)  :", vel_meta.get('Y_STEP'))
```
