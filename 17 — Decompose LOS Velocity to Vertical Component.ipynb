17 — Decompose LOS Velocity to Vertical Component

**What it does:** Converts Line-of-Sight (LOS) displacement rates into the true
vertical (up/down) component by dividing by the cosine of the satellite incidence angle.
This is the standard first-order decomposition assuming negligible horizontal motion.  
**When to use:** Essential for subsidence studies — your raw MintPy velocity is in LOS
(a slant direction), not vertical. For Dhaka or any flat alluvial plain where
horizontal motion is small, this single-step decomposition is valid and widely published.


```python
import numpy as np
from mintpy.utils import readfile

vel_path = f"{DATA_DIR}/velocity.h5"
velocity, meta = readfile.read(vel_path)

# Incidence angle from metadata (degrees → radians)
inc_angle_deg = float(meta.get('INCIDENCE_ANGLE', 33.0))   # Sentinel-1 ~ 33–46°
inc_angle_rad = np.deg2rad(inc_angle_deg)

# Vertical = LOS / cos(incidence_angle)
vertical_velocity = velocity / np.cos(inc_angle_rad)

unit = meta.get('UNIT', 'cm/year')
print(f"Incidence angle used       : {inc_angle_deg:.1f}°")
print(f"LOS mean velocity          : {np.nanmean(velocity):.3f} {unit}")
print(f"Vertical mean velocity     : {np.nanmean(vertical_velocity):.3f} {unit}")
print(f"Note: negative = subsidence, positive = uplift")
```
