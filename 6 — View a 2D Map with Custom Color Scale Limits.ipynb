6 — View a 2D Map with Custom Color Scale Limits
What it does: Renders the velocity (or any 2D file) as a map but clamps the color scale to your chosen min/max range so subtle deformation patterns are not washed out.
When to use: Generating clean, standardized maps for presentations, papers, or reports. Always set --vlim symmetrically around zero for deformation maps.

```python


from mintpy.cli import view

file_path = f"{DATA_DIR}/velocity.h5"

# -v : color scale min and max
# -c : Matplotlib colormap name
view.main([file_path, '-v', '-2', '2', '-c', 'RdBu'])
```
