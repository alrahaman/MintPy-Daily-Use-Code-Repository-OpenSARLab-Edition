13 — Filter Out Noisy Pixels Using a Temporal Coherence Threshold

**What it does:** Masks out low-quality pixels (water surfaces, layover zones, decorrelated
areas) based on temporal coherence, replacing them with `NaN` so they are invisible in plots.  
**When to use:** Always run this before generating any final output map.
A coherence threshold of 0.70 is standard; raise to 0.75–0.80 for stricter quality in urban areas.

```python
import numpy as np
from mintpy.utils import readfile

vel_path = f"{DATA_DIR}/velocity.h5"
coh_path = f"{DATA_DIR}/temporalCoherence.h5"

velocity,  _ = readfile.read(vel_path)
coherence, _ = readfile.read(coh_path)

COHERENCE_THRESHOLD = 0.70   # adjust as needed

filtered_velocity = np.where(coherence >= COHERENCE_THRESHOLD, velocity, np.nan)

valid_pixels = np.sum(~np.isnan(filtered_velocity))
total_pixels = filtered_velocity.size
print(f"Coherence threshold : {COHERENCE_THRESHOLD}")
print(f"Valid pixels kept   : {valid_pixels:,} / {total_pixels:,}  ({100*valid_pixels/total_pixels:.1f}%)")
```
